Soal:
```
3x + 2y -  z =  1
 x - 3y + 2z = 17
2x +  y + 4z = 20
```

````python
A = [
    [3, 2, -1],
    [1, -3, 2],
    [2, 1, 4],
]
b = [1, 17, 20]
print("A =", A)
print("b =", b)
````

````python
M = [A[i][:] + [b[i]] for i in range(3)]

print("Matriks augmented:")
for baris in M:
    print(baris)
````

````python
def pilih_pivot(M, k):
    baris_terbesar = k
    for i in range(k + 1, 3):
        if abs(M[i][k]) > abs(M[baris_terbesar][k]):
            baris_terbesar = i
    if baris_terbesar != k:
        M[k], M[baris_terbesar] = M[baris_terbesar], M[k]
    return M
````

````python
def eliminasi(M):
    for k in range(2):
        M = pilih_pivot(M, k)
        if abs(M[k][k]) < 1e-12:
            return None
        for i in range(k + 1, 3):
            pengali = M[i][k] / M[k][k]
            for j in range(k, 4):
                M[i][j] -= pengali * M[k][j]
    if abs(M[2][2]) < 1e-12:
        return None
    return M

M = eliminasi(M)
if M is None:
    print("Sistem tidak punya jawaban tunggal.")
else:
    print("Setelah eliminasi (segitiga atas):")
    for baris in M:
        print([round(v, 4) for v in baris])
````

````python
if M is not None:
    z = M[2][3] / M[2][2]
    y = (M[1][3] - M[1][2] * z) / M[1][1]
    x = (M[0][3] - M[0][1] * y - M[0][2] * z) / M[0][0]
    print(f"x = {x:.4f}")
    print(f"y = {y:.4f}")
    print(f"z = {z:.4f}")
````

````python
if M is not None:
    for i in range(3):
        ruas_kiri = A[i][0] * x + A[i][1] * y + A[i][2] * z
        cocok = abs(ruas_kiri - b[i]) < 1e-9
        print(f"Persamaan {i+1}: {ruas_kiri:.4f} = {b[i]} ->", "COCOK" if cocok else "TIDAK COCOK")
````
