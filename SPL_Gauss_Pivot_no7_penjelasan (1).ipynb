**Cell 1 - Soal (input).** `A` menyimpan angka di depan x, y, z dari tiga persamaan, sedangkan `b` menyimpan angka di ruas kanan. Untuk mengganti soal, cukup ubah angka di cell ini. `print` dipakai untuk memastikan data terbaca benar.

````python
# CELL 1 - SOAL (INPUT)
# Koefisien x, y, z dari ketiga persamaan (satu baris = satu persamaan)
A = [
    [3, 2, -1],     # 3x + 2y - 1z
    [1, -3, 2],     # 1x - 3y + 2z
    [2, 1, 4],      # 2x + 1y + 4z
]
# Angka di ruas kanan tiap persamaan
b = [1, 17, 20]
print("A =", A)
print("b =", b)
````

**Cell 2 - Matriks augmented.** Koefisien dan konstanta digabung menjadi satu tabel 3 baris x 4 kolom. Kolom ke-4 adalah ruas kanan. Dengan begitu, setiap operasi pada baris otomatis ikut mengubah angka di ruas kanan. Tanda `[:]` membuat salinan agar `A` asli tidak ikut rusak.

````python
# CELL 2 - MATRIKS AUGMENTED (3 baris x 4 kolom)
# Tempelkan b sebagai kolom ke-4 pada tiap baris A
# A[i][:] = salinan baris, supaya A asli tidak berubah
M = [A[i][:] + [b[i]] for i in range(3)]

print("Matriks augmented:")
for baris in M:
    print(baris)
````

**Cell 3 - Memilih pivot.** Pivot adalah angka acuan yang dipakai untuk menolkan angka di bawahnya. Fungsi ini mencari baris dengan angka terbesar (`abs` = abaikan tanda minus) pada kolom `k`, lalu menukarnya ke posisi atas. Memilih angka terbesar membuat hitungan lebih stabil dan mengurangi kesalahan pembulatan. Cell ini hanya mendefinisikan fungsi, jadi belum menampilkan apa-apa.

````python
# CELL 3 - MEMILIH PIVOT (angka terbesar, tanda diabaikan)
def pilih_pivot(M, k):
    baris_terbesar = k                          # anggap baris k dulu yang terbesar
    for i in range(k + 1, 3):                   # periksa baris-baris di bawahnya
        if abs(M[i][k]) > abs(M[baris_terbesar][k]):
            baris_terbesar = i                  # simpan baris yang angkanya lebih besar
    if baris_terbesar != k:
        M[k], M[baris_terbesar] = M[baris_terbesar], M[k]   # tukar posisi dua baris
    return M
````

**Cell 4 - Eliminasi.** Untuk kolom pertama lalu kolom kedua: (1) pilih pivot terbesar, (2) jika pivot 0 maka berhenti karena sistem tidak punya jawaban tunggal, (3) hitung `pengali` = angka di bawah pivot dibagi pivot, lalu (4) kurangkan baris itu dengan `pengali x baris pivot` sehingga angkanya menjadi 0. Hasil akhirnya tabel berbentuk segitiga atas. `1e-12` dipakai karena komputer menyimpan pecahan dengan sedikit pembulatan, sehingga angka yang sangat kecil dianggap nol.

````python
# CELL 4 - ELIMINASI (membuat segitiga: angka di bawah diagonal jadi 0)
def eliminasi(M):
    for k in range(2):                          # kolom 1 dan kolom 2
        M = pilih_pivot(M, k)                   # langkah 3: pilih pivot terbesar
        if abs(M[k][k]) < 1e-12:                # langkah 4: pivot = 0 -> berhenti
            return None
        for i in range(k + 1, 3):               # langkah 5: tiap baris di bawah pivot
            pengali = M[i][k] / M[k][k]         # angka di bawah pivot dibagi pivot
            for j in range(k, 4):
                M[i][j] -= pengali * M[k][j]    # baris_i dikurangi pengali x baris pivot
    if abs(M[2][2]) < 1e-12:                    # pivot terakhir juga tidak boleh 0
        return None
    return M

M = eliminasi(M)
if M is None:
    print("Sistem tidak punya jawaban tunggal.")
else:
    print("Setelah eliminasi (segitiga atas):")
    for baris in M:
        print([round(v, 4) for v in baris])
````

**Cell 5 - Substitusi mundur.** Setelah berbentuk segitiga, baris terakhir hanya berisi `z`, jadi `z` langsung ketemu. Nilai `z` dipakai di baris kedua untuk mencari `y`. Lalu `y` dan `z` dipakai di baris pertama untuk mencari `x`. Itulah sebabnya hitungannya dari bawah ke atas. `:.4f` membatasi tampilan menjadi 4 angka desimal.

````python
# CELL 5 - SUBSTITUSI MUNDUR (hitung dari baris terakhir ke atas)
if M is not None:
    z = M[2][3] / M[2][2]                               # baris 3: hanya ada z
    y = (M[1][3] - M[1][2] * z) / M[1][1]               # baris 2: masukkan z, cari y
    x = (M[0][3] - M[0][1] * y - M[0][2] * z) / M[0][0] # baris 1: masukkan y dan z, cari x
    print(f"x = {x:.4f}")
    print(f"y = {y:.4f}")
    print(f"z = {z:.4f}")
````

**Cell 6 - Pengecekan.** Nilai x, y, z dimasukkan kembali ke tiga persamaan awal. Jika angka di ruas kiri sama dengan `b`, program menulis COCOK, artinya jawabannya benar. Ini cara cepat memastikan perhitungan tidak salah.

````python
# CELL 6 - PENGECEKAN JAWABAN
# Masukkan x, y, z ke persamaan awal; hasilnya harus sama dengan b
if M is not None:
    for i in range(3):
        ruas_kiri = A[i][0] * x + A[i][1] * y + A[i][2] * z
        cocok = abs(ruas_kiri - b[i]) < 1e-9
        print(f"Persamaan {i+1}: {ruas_kiri:.4f} = {b[i]} ->", "COCOK" if cocok else "TIDAK COCOK")
````
